# 📧 Spam Mail Detection using Machine Learning and NLP

## Project Overview
This project builds a Spam Mail Detection System using Machine Learning. The model classifies emails as Spam or Ham using TF-IDF Vectorization and Logistic Regression.

## 📚 Importing Required Libraries

In this section, we import all the required libraries for data processing, feature extraction, model training, and evaluation.

```python
# Importing important dependencies
import numpy as np # Because machine is able to understand better whwnever it is converted into array
import pandas as pd # for converting data into dataframe
```

```python
# Importing LogisticRegression Model(for Binary output or used for Classification tasks)
# sklearn- specially used for machine learning models.Ever single thing based on ML is present in 'sklearn'
'''
sklearn → Scikit-learn library
linear_model → Module containing linear algorithms
LogisticRegression → Specific algorithm imported from that module
'''
from sklearn.linear_model import LogisticRegression
```

```python
# Importing train test split function to split the data into training and testing data
from sklearn.model_selection import train_test_split
```

```python
# Import accurracy score to check the accuracy of the model
from sklearn.metrics import accuracy_score
```

```python
# To convert all the string format into numeric
from sklearn.feature_extraction.text import TfidfVectorizer
```

## 📂 Loading the Dataset

The dataset contains email messages labeled as Spam or Ham.

```python
df = pd.read_csv('mail_data.csv')
df
```

## 🔍 Exploratory Data Analysis (EDA)

Before training the model, we explore the dataset structure, check for missing values, and understand class distribution.

```python
# Information about the dataset
df.info()
```

```python
# Checking null values
df.isnull().sum().reset_index()
```

```python
# Basic Analysis to understand the data
# Checking the ham/spam count in the dataset
df['Category'].value_counts().reset_index()
```

## 🛠 Data Preprocessing

The categorical labels are converted into numerical values and the dataset is prepared for machine learning.

```python
# Checking unique values in the dataset
pd.unique(df['Category'])
```

```python
# Label Encoding
# Labling spam email as 0 and ham email as 1
df.loc[df['Category'] == 'ham','Category'] = 1
df.loc[df['Category'] == 'spam','Category'] = 0
```

```python
# Seperating the data into input data and Label(Output data)
X = df['Message']
Y = df['Category']
```

## ✂️ Splitting the Dataset

The dataset is divided into training and testing sets to evaluate model performance on unseen data.

```python
# Splitting the data into training and testing data
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2)
print(X.shape, X_train.shape, X_test.shape)
print(Y.shape, Y_train.shape, Y_test.shape)
```

## 🔤 TF-IDF Feature Extraction

Machine learning algorithms cannot directly process text data. TF-IDF converts text into numerical feature vectors.

```python
# Convert text data into numerical features using TF-IDF
# Machine learning models cannot understand text directly,
# so we transform text into numbers.

feature_extraction = TfidfVectorizer(
    min_df=1,            # Include words that appear in at least 1 document
    stop_words='english',# Remove common English words like "the", "is", "and"
    lowercase=True       # Convert all text to lowercase before processing
)
```

```python
# Learn vocabulary from training data and convert text to TF-IDF features
X_train_num = feature_extraction.fit_transform(X_train)

# Convert test data using the same learned vocabulary
X_test_num = feature_extraction.transform(X_test)
```

```python
# converting Y_train and Y_test values as intege
Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

## 🤖 Model Training

A Logistic Regression model is trained using the transformed training data.

```python
# Training the model
model = LogisticRegression()
```

```python
# Training the Logistic Regression model on the training data
model.fit(X_train_num, Y_train)
```

## 📊 Model Evaluation

The trained model is evaluated using accuracy scores on both training and testing datasets.

```python
# Evaluating the trained model
X_train_prediction = model.predict(X_train_num)
# Finding the accuracy score on the training data
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
print('Accuracy score of the training data : ', training_data_accuracy)
```

```python
# Evaluating the test data
X_test_prediction = model.predict(X_test_num)
# Finding the accuracy score on the test data
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
print('Accuracy score of the test data : ', test_data_accuracy)
```

## 📧 Spam Mail Prediction

The trained model is used to classify new email messages as Spam or Ham.

```python
# Build a predictive system
input_data = ["Congratulations! You've won a free ticket to the Bahamas! Click here to claim now."]
input_data_num = feature_extraction.transform(input_data)
```

```python
# Making prediction
prediction = model.predict(input_data_num)
if prediction == 1:
    print('Ham mail')
else:
    print('Spam mail')
```
